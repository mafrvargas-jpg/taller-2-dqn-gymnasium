## 2. Exploración inicial del ambiente

En esta sección se crea el ambiente LunarLander-v3 y se inspeccionan sus principales características, incluyendo el espacio de observaciones, el espacio de acciones, la forma y el tipo de los datos recibidos.

```python
import gymnasium as gym
import numpy as np

env = gym.make("LunarLander-v3")

print("Espacio de observaciones:")
print(env.observation_space)

print("\nEspacio de acciones:")
print(env.action_space)

observation, info = env.reset(seed=42)

print("\nObservación inicial:")
print(observation)

print("\nForma de la observación:")
print(observation.shape)

print("\nTipo de dato:")
print(observation.dtype)

env.close()
```

## 3. Espacio de observaciones y acciones

### 3.1 Prueba de las acciones

Se realiza una prueba individual de las cuatro acciones disponibles en el ambiente para observar la recompensa obtenida y verificar si la ejecución produce la terminación o truncamiento del episodio.

```python
import gymnasium as gym

env = gym.make("LunarLander-v3")

observation, info = env.reset(seed=42)

print("=== INFORMACIÓN DEL AMBIENTE ===")

print("\nEspacio de observaciones:")
print(env.observation_space)

print("\nEspacio de acciones:")
print(env.action_space)

print("\nNúmero de acciones:")
print(env.action_space.n)

print("\nAcciones disponibles:")
print("0 = No hacer nada")
print("1 = Motor de orientación izquierdo")
print("2 = Motor principal")
print("3 = Motor de orientación derecho")

print("\n=== PRUEBA DE ACCIONES ===")

for action in range(env.action_space.n):

    observation, reward, terminated, truncated, info = env.step(action)

    print(f"\nAcción {action}:")
    print("Recompensa:", reward)
    print("Terminated:", terminated)
    print("Truncated:", truncated)

    if terminated or truncated:
        env.reset()

env.close()
```

## 4. Variables y rangos de las observaciones

Se identifican las ocho variables que conforman el estado observado por el agente, junto con sus rangos y tipos de datos.

```python
import gymnasium as gym

env = gym.make("LunarLander-v3")

observation, info = env.reset(seed=42)

print("=== VARIABLES DE LA OBSERVACIÓN ===")

variables = [
    "Posición X",
    "Posición Y",
    "Velocidad X",
    "Velocidad Y",
    "Ángulo",
    "Velocidad angular",
    "Contacto pierna izquierda",
    "Contacto pierna derecha"
]

for i, variable in enumerate(variables):
    print(f"{i}: {variable}")
    print(f"   Valor: {observation[i]}")
    print(f"   Tipo: {type(observation[i])}")

print("\n=== RANGOS ===")

low = env.observation_space.low
high = env.observation_space.high

for i, variable in enumerate(variables):
    print(f"{i}: {variable} → [{low[i]}, {high[i]}]")

env.close()
```

## 5. Condiciones de terminación del episodio

Se verifica el comportamiento de las variables `terminated` y `truncated` proporcionadas por la API de Gymnasium al finalizar un episodio.

```python
import gymnasium as gym

env = gym.make("LunarLander-v3")

observation, info = env.reset(seed=42)

terminated = False
truncated = False
steps = 0

while not (terminated or truncated):

    action = env.action_space.sample()

    observation, reward, terminated, truncated, info = env.step(action)

    steps += 1

print("=== FIN DEL EPISODIO ===")
print("Número de pasos:", steps)
print("Terminated:", terminated)
print("Truncated:", truncated)

env.close()
```

## 6. Conclusiones de la exploración

El ambiente LunarLander-v3 presenta un espacio de observaciones continuo compuesto por 8 variables de tipo `float32`, relacionadas con la posición, velocidad, orientación y contacto de las piernas de la nave.

El espacio de acciones es discreto y contiene 4 acciones posibles: no realizar ninguna acción, activar el motor de orientación izquierdo, activar el motor principal o activar el motor de orientación derecho.

Las pruebas realizadas muestran que la recompensa depende tanto del estado de la nave como de la acción seleccionada. Por esta razón, el agente debe aprender una política que permita seleccionar la acción más conveniente en cada estado para maximizar la recompensa acumulada.

La estructura de observaciones y acciones del ambiente es compatible con un agente Deep Q-Network (DQN), ya que el estado puede representarse como un vector de características y las acciones corresponden a un conjunto discreto de alternativas.
